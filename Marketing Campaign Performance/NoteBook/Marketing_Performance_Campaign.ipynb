Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load Dataset

```python
df = pd.read_csv("marketing_campaign_dataset.csv")
df.head()
```

```python
df.info()
```

```python
#Change Data Type for Acquisition_Cost and #Date
df['Acquisition_Cost'] = df['Acquisition_Cost'].replace('[\$,]', '', regex=True).astype(float)
df['Date'] = pd.to_datetime(df['Date'])
```

```python
#Recheck data information again
df.info()
```

```python
df.describe()
```

Data Preprocessing and Cleaning

```python
#check for missing value
df.isnull().sum()
```

```python
#Check for duplicates
df.duplicated().sum()
```

```python
#Creating Month Column
df['Month'] = df['Date'].dt.month
```

```python
df.head()
```

#KPI Calculations

```python
#Click Through Rate
df["CTR"] = df["Clicks"] / df["Impressions"]
```

```python
#Number of Conversions
df["Conversions"] = df["Clicks"] * df["Conversion_Rate"]
```

```python
#Revenue
df["Revenue"] = df["Acquisition_Cost"] * (1 + df["ROI"])
```

```python
#Calculate Return on Ad Spend
df["ROAS"] = df["Revenue"] / df["Acquisition_Cost"]
```

```python
df.head()
```

#Exploratory Data Analysis (EDA)

```python
#Channel Performance
df.groupby("Channel_Used")[["Revenue","Acquisition_Cost"]].sum()
```

```python
# ROI by Channel
df.groupby("Channel_Used")["ROI"].mean().sort_values(ascending=False)
```

```python
#Conversion Rate by Channel
df.groupby("Channel_Used")["Conversion_Rate"].mean()
```

```python
# Top Performing Campaigns
df.sort_values("ROI", ascending=False).head(10)
```

```python
# Worst Campaigns
df.sort_values("ROI").head(10)
```

#Export Clean Dataset

```python
df.to_csv("marketing_campaign_cleaned.csv", index=False)
```
